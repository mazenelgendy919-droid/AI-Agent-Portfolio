LLM Engine Initialization (4-bit Quantization)

```python
# Cell 2: LLM Engine Initialization (4-bit Quantization)
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig, pipeline
from langchain_huggingface import HuggingFacePipeline

print("⏳ Loading Model... This might take a few minutes for a 7B model.")

# 1. Configure 4-bit Quantization to save VRAM
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4"
)

# 2. Load the highly capable Qwen2.5 7B Instruct Model
model_id = "Qwen/Qwen2.5-7B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto"
)

# 3. Create a text-generation pipeline
text_pipeline = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=256,
    temperature=0.3, # Low temperature for professional, focused email writing
    repetition_penalty=1.1,
    return_full_text=False
)

# 4. Wrap with LangChain for seamless integration
llm = HuggingFacePipeline(pipeline=text_pipeline)

print("✅ Cell 2: Qwen2.5-7B (4-bit Quantized) loaded successfully!")
```

Backend Automation (API-Like Functions)

```python
# Cell 3: Backend Automation (Mock API Functions)

def api_log_interaction(customer_email: str, ai_reply: str) -> None:
    """Simulates an API endpoint logging the interaction to a monitoring system."""
    log_data = {
        "timestamp": datetime.now().isoformat(),
        "customer_message": customer_email,
        "ai_response": ai_reply
    }
    logging.info(f"[API: LOG] Interaction logged successfully. Length of reply: {len(ai_reply)} chars.")

def api_save_to_database(email_content: str, filename="email_records.json") -> bool:
    """Simulates an API endpoint saving the email record to a database/file."""
    try:
        record = {"date": datetime.now().isoformat(), "content": email_content}

        # In a real scenario, this connects to SQL/NoSQL. Here we use a JSON file.
        if not os.path.exists(filename):
            with open(filename, 'w') as f:
                json.dump([record], f)
        else:
            with open(filename, 'r+') as f:
                data = json.load(f)
                data.append(record)
                f.seek(0)
                json.dump(data, f, indent=4)

        logging.info("[API: SAVE] Email saved to database successfully.")
        return True
    except Exception as e:
        logging.error(f"[API: SAVE ERROR] Failed to save email: {str(e)}")
        raise Exception(f"Database save failed: {str(e)}")

def api_send_email(recipient: str, subject: str, body: str) -> bool:
    """Simulates an API endpoint sending an email via SMTP/SendGrid."""
    logging.info(f"[API: SEND] Sending email to '{recipient}' with subject '{subject}'...")
    # Simulating network delay
    import time
    time.sleep(1)
    logging.info("[API: SEND] Email sent successfully!")
    return True

print("✅ Cell 3: API automation functions initialized.")
```

Core Object-Oriented System (The AI Assistant)

```python
# Cell 4: Core System Architecture (Automated Email Assistant)
from langchain_core.prompts import PromptTemplate

class AutomatedEmailAssistant:
    def __init__(self, llm_engine):
        self.llm = llm_engine
        # Define a strict prompt to ensure structural email formatting
        self.prompt_template = PromptTemplate.from_template(
            "You are a professional customer support agent for a tech company.\n"
            "Customer Message: {customer_message}\n"
            "Task: Write a polite, professional, and helpful email reply to the customer.\n"
            "Start directly with 'Dear Customer,' and end with 'Best regards, Tech Support'.\n"
            "Reply:"
        )

    def generate_reply(self, customer_message: str) -> str:
        """Generates the AI response."""
        chain = self.prompt_template | self.llm
        return chain.invoke({"customer_message": customer_message}).strip()

    def process_incoming_email(self, customer_message: str, customer_email: str = "customer@example.com"):
        """The main automation pipeline."""
        print(f"\n--- Processing New Message from {customer_email} ---")

        # 1. Error Handling: Check for empty input
        if not customer_message or not customer_message.strip():
            logging.warning("Pipeline Halted: Empty customer message received.")
            return {"status": "error", "reason": "Empty input"}

        try:
            # 2. AI Generation
            print("🤖 AI is drafting the reply...")
            ai_reply = self.generate_reply(customer_message)

            if not ai_reply:
                raise ValueError("AI generated an empty response.")

            # 3. API Automation: Logging
            api_log_interaction(customer_message, ai_reply)

            # 4. API Automation: Saving to DB
            api_save_to_database(email_content=ai_reply)

            # 5. API Automation: Sending Email
            api_send_email(
                recipient=customer_email,
                subject="Re: Your Customer Support Inquiry",
                body=ai_reply
            )

            print("\n📬 FINAL EMAIL DRAFT:")
            print("="*50)
            print(ai_reply)
            print("="*50)

            return {"status": "success", "reply": ai_reply}

        except Exception as e:
            # Error Handling: Catching unexpected API or System failures
            logging.error(f"Pipeline Failed: Critical error during processing - {str(e)}")
            return {"status": "error", "reason": str(e)}

print("✅ Cell 4: AutomatedEmailAssistant Class is ready.")
```

Pipeline Execution & Error Testing

```python
# Cell 5: Execution, Automation Simulation, and Error Handling Tests

# Initialize the assistant with our quantized local model
assistant = AutomatedEmailAssistant(llm_engine=llm)

print("\n" + "#"*60)
print("🧪 SCENARIO 1: Valid Customer Message (Happy Path)")
print("#"*60)
customer_msg_1 = "Hello, I recently bought a laptop from your store, but the screen is flickering. How can I claim my warranty?"
assistant.process_incoming_email(customer_message=customer_msg_1, customer_email="ahmed@gmail.com")


print("\n" + "#"*60)
print("🧪 SCENARIO 2: Empty Input Error Handling (Edge Case)")
print("#"*60)
# Testing the system's ability to handle empty inputs gracefully without crashing
customer_msg_2 = "   "
assistant.process_incoming_email(customer_message=customer_msg_2, customer_email="spam@gmail.com")
```
